# What problem are we trying to solve?
We are trying to use attributes of flowers to predict the species of the flowers. Specifically, we are trying to use the sepal length and width and the petal length and width to predict if an Iris flower is Setosa, Versicolor, or Virginica

# Some Exploratory Data Analysis With Iris

```python
fig, axes = plt.subplots(2, 2, figsize=(10, 8))

data.SepalLengthCm.plot.hist(ax=axes[0, 0], title='Sepal Length')
data.SepalWidthCm.plot.hist(ax=axes[0, 1], title='Sepal Width')
data.PetalLengthCm.plot.hist(ax=axes[1, 0], title='Petal Length')
data.PetalWidthCm.plot.hist(ax=axes[1, 1], title='Petal Width')

plt.tight_layout()
plt.show()
```

As we can see that Petal features are quite distinctive compared to Sepal features. So, Petal features can be used to distinguish in between species
